```python
class Dog:
    def __init__(self, name, breed, owner):
        self.name = name
        self.breed = breed
        self.owner = owner

    def bark(self):
        print(f"my name is {self.name} and i am a {self.breed}, owner name is {self.owner}")

class Owner:
    def __init__(self, name, address, contact_no):
        self.name = name
        self.address = address
        self.contact_no = contact_no

owner1 = Dog("mike", "pomaranian", "hari")
owner1.bark()
```

```python
class Person:
    def __init__(self, name, fav_club):
        self.name = name
        self.fav_club = fav_club

    def info(self):
        print(f"my name is {self.name} and i support {self.fav_club}")

summa = Person("hari", "Liverpool FC")
summa.info()
```
