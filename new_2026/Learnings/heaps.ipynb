Heaps

```python
#Build Min Heap (Heapify)

A = [-4, 3, 1, 0, 2, 5, 10, 8, 12, 9]

import heapq
heapq.heapify(A)

print(A)
```

```python
# heap push

heapq.heappush(A, 3)
print(A)
```

```python
# heap pop

minn = heapq.heappop(A)
print(A, minn)
```

```python
# heap sort

def heapsort(arr):
    heapq.heapify(arr)
    n = len(arr)
    new_list = [0] * n

    for i in range(n):
        minn = heapq.heappop(arr)
        new_list[i] = minn

    return new_list

heapsort([1, 3, 5, 7, 9, 2, 4, 6, 8, 0])
```

```python
# max heap

B = [-4, 3, 1, 0, 2, 5, 10, 8, 12, 9]
n = len(B)

for i in range(n):
    B[i] = -B[i]

heapq.heapify(B)

print(B)

largest = -heapq.heappop(B)
print(largest)
```

```python
# while inserting also we need to use minus.

heapq.heappush(B,-7)
print(B)
```

```python
# building heap from scratch

C = [-5, 4, 2, 1, 7, 0, 3]
heap = []

for x in C:
    heapq.heappush(heap, x)
    print(heap, len(heap))
```
