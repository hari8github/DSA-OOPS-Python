Dynamic programming - Fibonacci

Naive recursion

```python
def fib_1(n):
    if n <= 1:
        return n
    return fib_1(n-1) + fib_1(n-2)

fib_1(6)
```

Top Down - memoization

```python
def fib_2(n):
    memo = {}
    if n <= 1:
        return n
    if n in memo:
        return memo[n]
    memo[n] = fib_2(n-1) + fib_2(n-2)
    return memo[n]

fib_2(6)
```

Bottom up - tabulation

```python
def fib_3(n):
    if n <= 1:
        return n
    dp = [0] * (n+1)
    dp[1] = 1
    for i in range(2, n+1):
        dp[i] = fib_3(i-1) + fib_3(i-2)
    return dp[n]

fib_3(6)
```

Bottom up - constant time

```python
def fib_4(n):
    a, b = 0,1
    for _ in range(2, n+1):
        a,b = b, a+b
    return b

fib_4(6)
```
