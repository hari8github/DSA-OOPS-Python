Sliding window - Fixed length

Given an array and integer k, find the subarray of length k with the maximum average. Return that average.

```python
def max_avg(arr, k):
    n = len(arr)
    window_sum = sum(arr[:k])
    max_sum = 0

    for i in range(k, n):
        window_sum += arr[i] - arr[i-k]
        max_sum = max(max_sum, window_sum)

    return max_sum / k

nums = [1, 12, -5, -6, 50, 3]
print(max_avg(nums, 4))
```

Sliding Window - variable length

Find the length of the longest substring where no character repeats.

```python
def longest_substring(str):
    seen = set()
    l = 0
    max_len = 0

    for r in range(len(str)):
        while str[r] in seen:
            seen.remove(str[l])
            l += 1
        seen.add(str[r])
        max_len = max(max_len, r-l+1)
    return max_len

print(longest_substring("abcabcbbc"))
```
