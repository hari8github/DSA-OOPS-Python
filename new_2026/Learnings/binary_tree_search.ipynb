Binary Trees

```python
class TreeNode:
    def __init__(self, val, right = None, left = None):
        self.val = val
        self.right = right
        self.left = left

    def __str__(self):
        return str(self.val)
```

```python
#        1
#     2    3
#   4  5  10
```

```python
A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(10)

A.left = B
A.right = C
B.left = D
B.right = E
C.left = F

print(A)
```

```python
# Recursive Pre Order Traversal (DFS)

def pre_order(node):
    if not node:
        return
    print(node)
    pre_order(node.left)
    pre_order(node.right)

pre_order(A)
```

```python
# Recursive In order traversal (DFS)

def in_order(node):
    if not node:
        return 
    in_order(node.left)
    print(node)
    in_order(node.right)

in_order(A)
```

```python
# Recursive Post order traversal (DFS)

def post_order(node):
    if not node:
        return 
    post_order(node.left)
    post_order(node.right)
    print(node)

post_order(A)
```

```python
# Iterative Pre Order (DFS)

def pre_order_iterative(node):
    stk = [node]

    while stk:
        node = stk.pop()
        if node.right: stk.append(node.right)
        if node.left: stk.append(node.left)
        print(node)
pre_order_iterative(A)
```

```python
# LEVEL order traversal (BFS)

from collections import deque

def level_order(node):
    q = deque()
    q.append(node)

    while q:
        node = q.popleft()
        print(node)
        if node.left: q.append(node.left)
        if node.right: q.append(node.right)
level_order(A)
```

```python
# to check if the value exists DFS

def search(node, target):
    if not node:
        return False
    if node.val == target:
        return True
    return search(node.left, target) or search(node.right, target)

search(A, 3)
```

```python
# Binary Search Trees (BSTs)

#       5
#    1    8
#  -1 3  7 9
```

```python
A2 = TreeNode(5)
B2 = TreeNode(1)
C2 = TreeNode(8)
D2 = TreeNode(-1)
E2 = TreeNode(3)
F2 = TreeNode(7)
G2 = TreeNode(9)

A2.left, A2.right = B2, C2
B2.left, B2.right = D2, E2
C2.left, C2.right = F2, G2

print(A2)
```

```python
in_order(A2)
```

```python
# binary search tree (BST)

def search_bst(node, target):
    if not node:
        return False
    if node.val == target:
        return True
    if target < node.val : return search_bst(node.left, target)
    else: return search_bst(node.right, target)

search_bst(A2, 9)
```
