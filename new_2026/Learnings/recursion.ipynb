Recursion

```python
# fibonacci

def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

fib(3)
```

```python
# linkedlists - printing the elements in reverse order.

class Singlynode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next

    def __str__(self):
        return str(self.val)

head = Singlynode(1)
a = Singlynode(2)
b = Singlynode(3)

head.next = a
a.next = b

print(head)
```

```python
def reverse(node):
    if not node:
        return
    
    reverse(node.next)
    print(node)

reverse(head)
```
