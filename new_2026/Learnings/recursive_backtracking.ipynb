Finding all possible Subsets

```python
def subsets(nums):
    n = len(nums)
    res, sol = [], []

    def backtrack(i):
        if i == n :
            res.append(sol[:])
            return

        backtrack(i+1)
        sol.append(nums[i])
        backtrack(i+1)
        sol.pop()

    backtrack(0)
    return res
nums = [1,2]
print(subsets(nums))
```
