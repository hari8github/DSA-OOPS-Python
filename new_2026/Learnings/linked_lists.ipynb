Singly linked list

```python
class Singlynode:
    def __init__(self, val, next = None):
        self.val = val
        self.next = next

    def __str__(self):
        return str(self.val)
```

```python
Head = Singlynode(1)
A = Singlynode(2)
B = Singlynode(3)
C = Singlynode(7)

Head.next = A
A.next = B
B.next = C

print(Head)
```

```python
curr = Head

while curr:
    print(curr)
    curr = curr.next
```

```python
def display(head):
    curr = head
    elements = []
    while curr:
        elements.append(str(curr.val))
        curr = curr.next
    print(' -> '.join(elements))

display(Head)
```

```python
def search(head, val):
    curr = head
    while curr:
        if val == curr.val:
            return True
        curr = curr.next

    return False

search(Head, 4)
```

Doubly linked list

```python
class Doublynode():
    def __init__(self, val, next = None, prev = None):
        self.val = val 
        self.next = next
        self.prev = prev
    def __str__(self):
        return str(self.val)
```

```python
head = tail = Doublynode(1)
print(tail)
```

```python
def display(head):
    curr = head
    elements = []
    while curr:
        elements.append(str(curr.val))
        curr = curr.next
    print(' <-> '.join(elements))

display(head)
```

```python
def insert_at_the_beginning(head, tail, val):
    new_node = Doublynode(val, next = head)
    head.prev = new_node
    return new_node, tail

head, tail = insert_at_the_beginning(head, tail, 3)
display(head)
```

```python
def insert_at_end(head, tail, val):
    new_node = Doublynode(val, prev = tail)
    tail.next = new_node
    return head, new_node

head, tail = insert_at_end(head, tail, 6)
display(head)
```
