Binary Search

Traditional Binary Search

```python
def binary_search(arr, target):
    N = len(arr)
    L = 0
    R = N - 1

    while L <= R:
        M = L + ((R-L) // 2)
        if arr[M] == target:
            return True
        elif target < arr[M]:
            R = M - 1
        else:
            L = M + 1
    
    return False

nums = [1,3,5,7,9]
print(binary_search(nums, 5))
print(binary_search(nums, 2))
```

Condition Based

```python
summa = [False, True, True, True, True]

def binary_search_condition(arr):
    N = len(arr)
    L = 0
    R = N-1

    while L < R:
        M = (L+R) // 2
        if arr[M]:
            R = M
        else:
            L = M+1

    return L

print(binary_search_condition(summa))
```

```python
# Range-Based

def guess_number(target, L, R):
    while L <= R:
        M = (L + R) // 2
        if M == target:
            return M  # Found the number
        elif M < target:
            L = M + 1
        else:
            R = M - 1
    return -1  # Target not found


target = 900
result = guess_number(target, 1, 100)
print(f"The guessed number is: {result}")
```
