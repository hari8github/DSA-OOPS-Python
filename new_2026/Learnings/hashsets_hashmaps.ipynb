Hashsets

```python
s = set()
print(s)
```

```python
# adding items - o(1)
s.add(1)
s.add(2)
s.add(3)

print(s)
```

```python
# lookup - o(1)

if 4 not in s:
    print(True)
```

```python
# remove - o(1)

s.remove(3)
print(s)
```

```python
# set construction - o(s) - s = length of string

string = 'aaaabbbcccd'
sett = set(string)

print(sett) # sets are unordered. 
```

```python
# loop - o(n)
for x in s:
    print(x)
```

Hashmaps - Dictionaries

```python
d = {'greg' : 1, 'maths' : 2, 'football' : 3}
print(d)
```

```python
# add key:val pairs. - o(1)
d['scofield'] = 4
print(d)
```

```python
# lookup - o(1)
if 'scofield' in d:
    print(True)
```

```python
# check value to its key - o(1)
print(d['scofield'])
```

```python
# loop - o(n)
for key, val in d.items():
    print(f'KEY: ', key, 'VALUE: ', val)
```
