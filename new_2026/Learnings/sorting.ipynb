Bubble sort

```python
def bubble_sort(arr):
    n = len(arr)
    flag = True
    while flag:
        flag = False
        for i in range(1, n):
            if arr[i-1] > arr[i]:
                flag = True
                arr[i-1], arr[i] = arr[i], arr[i-1]

A = [-1, -2, 1, 6, 9, 8, 5]
bubble_sort(A)
print(A)
```

Insertion Sort

```python
def insertion_sort(arr):
    n = len(arr)
    for i in range(1,n):
        for j in range(i, 0, -1):
            if arr[j-1] > arr[j]:
                arr[j-1], arr[j] = arr[j] , arr[j-1]
            else:
                break
A = [-1, -2, 1, 6, 9, 8, 5]
insertion_sort(A)
A
```

Selection Sort

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if arr[j] < arr[i]:
                min_index = j
        arr[i] , arr[min_index] = arr[min_index], arr[i]

A = [-1, -2, 1, 6, 9, 8, 5]
selection_sort(A)
A
```

Merge sort

```python
def merge_sort(arr):
    n = len(arr)

    if n == 1:
        return arr

    m = len(arr) // 2
    L = arr[:m]
    R = arr[m:]

    L = merge_sort(L)
    R = merge_sort(R)

    l, r = 0, 0
    L_len = len(L)
    R_len = len(R)

    sorted_arr = [0] * n
    i = 0

    while l < L_len and r < R_len:
        if L[l] < R[r]:
            sorted_arr[i] = L[l]
            l += 1
        else:
            sorted_arr[i] = R[r]
            r += 1
        i+=1
        
    while l < L_len:
        sorted_arr[i] = L[l]
        l+=1
        i+=1
    while r < R_len:
        sorted_arr[i] = R[r]
        r += 1
        i += 1

    return sorted_arr

A = [-1, -2, 1, 6, 9, 8, 5]
print(merge_sort(A))
```

Quick sort

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    p = arr[-1]
    L = [x for x in arr[:-1] if x <= p]
    R = [x for x in arr[:-1] if x > p]

    L = quick_sort(L)
    R = quick_sort(R)

    return L + [p] + R
A = [-1, -2, 1, 6, 9, 8, 5]
quick_sort(A)
A
```
