```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def get_age(self):
        return self.age

    def set_age(self, age):
        if age > 0:
            self.age = age
        else:
            print("Age must be greater than 0")

d = Dog("sheldon", 34)
print(d.get_name())

print(d.get_age())

age = d.set_age(0)
print(d.get_age())
```

```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks


    def get_marks(self):
        return self.marks

class Course:
    def __init__(self, name, max_stds):
        self.name = name
        self.max_stds = max_stds
        self.students = []

    # Adding a method to add students to the course where if the length of students is less than max_stds, the student is added.

    def add_students(self, student):
        if len(self.students) < self.max_stds:
            self.students.append(student)
            return True
        return False

    def get_avg_marks(self):
        value = 0
        for std in self.students:
            value += std.get_marks()
        return value / len(self.students)

s1 = Student("hari", 21, 99)
s2 = Student("venkat", 21, 99)
s3 = Student("ghost" , 21, 9)

c = Course("physics", 2)
c.add_students(s1)
c.add_students(s3)
print(c.get_avg_marks())
print(c.students)
print(c.students[1].name)

print(c.add_students(s2))
```

```python
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary

    def get_name(self):
        return self.name

    def get_department(self):
        return self.department
    
    def get_salary(self):
        return self.salary

    def set_salary(self, new_salary):
        if new_salary > 0:
            self.salary = new_salary
            print(f"Salary updated to {new_salary}")
        else:
            print("Salary must be positive")

class Company:
    def __init__(self, name):
        self.name = name
        self.employees = []

    def hire_employee(self, employee):
        """Add an employee to the company"""
        self.employees.append(employee)
        print(f"Hired {employee.get_name()} in {employee.get_department()}")

    def get_totakl_payroll(self):
        """Calculate total salary expenses"""
        total = 0
        for emp in self.employees:
            total += emp.get_salary()

        return total

    def get_employees_by_department(self, department):
        """Return list of employees in a specific department"""
        dept_employees = []
        for emp in self.employees:
            if emp.get_department() == department:
                dept_employees.append(emp)
        return dept_employees

if __name__ == "__main__":
    # Create company
    company = Company("Tech Corp")
    
    # Create employees
    emp1 = Employee("Hari", "Engineering", 75000)
    emp2 = Employee("Venkat", "Engineering", 80000)
    emp3 = Employee("Priya", "Marketing", 65000)
    
    # Hire employees
    company.hire_employee(emp1)
    company.hire_employee(emp2)
    company.hire_employee(emp3)
    
    # Give raises
    emp1.give_raise(10)  # 10% raise
    
    # Check payroll
    print(f"Total payroll: ${company.get_total_payroll()}")
    
    # Get engineering team
    eng_team = company.get_employees_by_department("Engineering")
    print(f"Engineering team has {len(eng_team)} members")
```
